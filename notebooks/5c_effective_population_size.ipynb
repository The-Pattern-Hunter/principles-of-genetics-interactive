# Part 1: The Concept - Why Ne < N?

## The Surprising Discovery

You count 1000 fish in a pond. But genetically, it's as if only 200 are contributing genes to the next generation!

**Why?**

---

## 🎓 Three Levels of Understanding

### 9th Grade: The Lottery Analogy

Imagine a village lottery for next year's farming committee:
- 1000 people in village (census size N = 1000)
- But only 100 people buy lottery tickets
- Only 50 actually win seats
- **Effective size for decision-making = 50, not 1000!**

Similarly in populations:
- Not all adults breed
- Some have many offspring, some have few
- Only **effective** breeders matter for genetics

### BSc Level: The Genetic Reality

**Effective population size (Ne):** The size of an **ideal** population that would have the same rate of genetic drift as the actual population.

**Ideal population:**
- All individuals breed
- Equal sex ratio
- Random mating
- Non-overlapping generations
- Equal family sizes (Poisson distributed)

**Real populations violate these!** So Ne < N.

### MSc/Research: The Mathematical Framework

**Rate of genetic drift:**

$$\sigma^2_{\Delta p} = \frac{p(1-p)}{2N_e}$$

**Ne is defined such that** this variance formula holds.

**Key equations:**

**Unequal sex ratio:**
$$N_e = \frac{4N_mN_f}{N_m + N_f}$$

**Variance in offspring number:**
$$N_e = \frac{4N - 2}{V_k + 2}$$

Where Vk = variance in offspring number

**References:**
- Wright, S. (1931). Evolution in Mendelian populations. *Genetics* 16:97-159.
- Crow, J.F. & Kimura, M. (1970). *An Introduction to Population Genetics Theory*.
- Charlesworth, B. (2009). Effective population size. *Current Biology* 19:R711-R712.

## Interactive 1: Explore Ne/N Ratio

**EDUCATIONAL SIMULATION**

See how various factors reduce Ne below census size N!

```python
def explore_ne_ratio(census_N=1000, males=400, females=600, 
                    breeding_fraction=0.5, offspring_variance=4):
    """
    Calculate Ne under different scenarios
    
    EDUCATIONAL TOOL - Demonstrates factors reducing Ne
    """
    
    # Calculate Ne components
    # 1. Sex ratio effect
    Nm_breeding = males * breeding_fraction
    Nf_breeding = females * breeding_fraction
    Ne_sexratio = (4 * Nm_breeding * Nf_breeding) / (Nm_breeding + Nf_breeding)
    
    # 2. Variance in reproductive success
    N_breeding = Nm_breeding + Nf_breeding
    Ne_variance = (4 * N_breeding - 2) / (offspring_variance + 2)
    
    # 3. Combined effect (approximate)
    Ne_combined = min(Ne_sexratio, Ne_variance)
    
    # Ratio
    Ne_N_ratio = Ne_combined / census_N
    
    # Visualization
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    
    # Plot 1: Population composition
    categories = ['Census N', 'Breeding\nMales', 'Breeding\nFemales', 'Effective\nNe']
    values = [census_N, Nm_breeding, Nf_breeding, Ne_combined]
    colors = ['gray', 'blue', 'pink', 'green']
    
    bars = axes[0,0].bar(categories, values, color=colors, alpha=0.6, 
                        edgecolor='black', linewidth=2)
    axes[0,0].set_ylabel('Number of Individuals', fontsize=12)
    axes[0,0].set_title('Census N vs Effective Ne', fontsize=14, weight='bold')
    axes[0,0].grid(True, alpha=0.3, axis='y')
    
    # Add value labels
    for bar, val in zip(bars, values):
        height = bar.get_height()
        axes[0,0].text(bar.get_x() + bar.get_width()/2., height + 20,
                      f'{int(val)}', ha='center', va='bottom', 
                      fontsize=11, weight='bold')
    
    # Plot 2: Ne/N ratio visualization
    # Draw a circle representing N
    circle_N = plt.Circle((0.5, 0.5), 0.4, color='gray', alpha=0.3, label=f'Census N={census_N}')
    axes[0,1].add_patch(circle_N)
    
    # Draw a smaller circle representing Ne
    radius_Ne = 0.4 * np.sqrt(Ne_N_ratio)
    circle_Ne = plt.Circle((0.5, 0.5), radius_Ne, color='green', alpha=0.6, label=f'Effective Ne={int(Ne_combined)}')
    axes[0,1].add_patch(circle_Ne)
    
    axes[0,1].set_xlim(0, 1)
    axes[0,1].set_ylim(0, 1)
    axes[0,1].set_aspect('equal')
    axes[0,1].set_title('Genetic Bottleneck Visualization', fontsize=14, weight='bold')
    axes[0,1].legend(loc='upper right', fontsize=10)
    axes[0,1].axis('off')
    
    # Add ratio text
    axes[0,1].text(0.5, 0.1, f'Ne/N = {Ne_N_ratio:.2f}', 
                  ha='center', fontsize=16, weight='bold',
                  bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))
    
    # Plot 3: Factors reducing Ne
    factors = ['Ideal\nPopulation', 'Sex Ratio\nEffect', 'Offspring\nVariance', 'Combined\nReality']
    Ne_values = [census_N * breeding_fraction, Ne_sexratio, Ne_variance, Ne_combined]
    colors_factors = ['lightgreen', 'lightblue', 'lightcoral', 'orange']
    
    bars2 = axes[1,0].bar(factors, Ne_values, color=colors_factors, alpha=0.7,
                         edgecolor='black', linewidth=2)
    axes[1,0].axhline(y=census_N, color='red', linestyle='--', linewidth=2, 
                     label=f'Census N = {census_N}', alpha=0.7)
    axes[1,0].set_ylabel('Effective Size', fontsize=12)
    axes[1,0].set_title('How Factors Reduce Ne', fontsize=14, weight='bold')
    axes[1,0].legend(fontsize=10)
    axes[1,0].grid(True, alpha=0.3, axis='y')
    
    # Add value labels
    for bar, val in zip(bars2, Ne_values):
        height = bar.get_height()
        axes[1,0].text(bar.get_x() + bar.get_width()/2., height + 20,
                      f'{int(val)}', ha='center', va='bottom', 
                      fontsize=10, weight='bold')
    
    # Plot 4: Genetic drift rate
    # Simulate drift with different Ne values
    generations = 50
    
    # With census N (hypothetical)
    p_census = [0.5]
    # With effective Ne (reality)
    p_effective = [0.5]
    
    np.random.seed(42)
    for gen in range(generations):
        # Census size drift
        p_new_census = np.random.binomial(2*int(census_N*breeding_fraction), p_census[-1]) / (2*census_N*breeding_fraction)
        p_census.append(p_new_census)
        
        # Effective size drift
        p_new_eff = np.random.binomial(2*int(Ne_combined), p_effective[-1]) / (2*Ne_combined)
        p_effective.append(p_new_eff)
    
    gens = np.arange(0, generations + 1)
    axes[1,1].plot(gens, p_census, 'g-', linewidth=2, label=f'If Ne = N×{breeding_fraction} (ideal)', alpha=0.7)
    axes[1,1].plot(gens, p_effective, 'r-', linewidth=2.5, label=f'With real Ne = {int(Ne_combined)}', alpha=0.8)
    axes[1,1].axhline(y=0.5, color='gray', linestyle=':', alpha=0.5)
    
    axes[1,1].set_xlabel('Generation', fontsize=12)
    axes[1,1].set_ylabel('Allele Frequency', fontsize=12)
    axes[1,1].set_title('Genetic Drift Rate', fontsize=14, weight='bold')
    axes[1,1].legend(fontsize=10)
    axes[1,1].grid(True, alpha=0.3)
    axes[1,1].set_ylim(0, 1)
    
    # Add annotation
    axes[1,1].text(0.02, 0.98, 'Lower Ne → FASTER drift!',
                  transform=axes[1,1].transAxes, fontsize=11,
                  va='top', bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
    
    # Interpretation
    print("\n" + "="*70)
    print("Ne CALCULATION AND INTERPRETATION")
    print("="*70)
    
    print(f"\nPopulation parameters:")
    print(f"  • Census size (N): {census_N}")
    print(f"  • Males: {males} ({males/census_N*100:.1f}%)")
    print(f"  • Females: {females} ({females/census_N*100:.1f}%)")
    print(f"  • Breeding fraction: {breeding_fraction*100:.1f}%")
    print(f"  • Offspring variance (Vk): {offspring_variance}")
    
    print(f"\nEffective population size:")
    print(f"  • Ne from sex ratio: {Ne_sexratio:.0f}")
    print(f"  • Ne from offspring variance: {Ne_variance:.0f}")
    print(f"  • Combined Ne: {Ne_combined:.0f}")
    print(f"  • Ne/N ratio: {Ne_N_ratio:.3f}")
    
    print(f"\nWhat does this mean?")
    print(f"  • Only {Ne_N_ratio*100:.1f}% of census size is 'genetically effective'")
    print(f"  • Genetic drift acts {census_N/Ne_combined:.1f}× FASTER than expected from N")
    print(f"  • Population loses genetic variation {census_N/Ne_combined:.1f}× faster")
    
    print(f"\nConservation implications:")
    if Ne_combined < 50:
        print("  ⚠️  CRITICAL: Ne < 50")
        print("  • Severe risk of inbreeding depression")
        print("  • Genetic rescue needed immediately")
        print("  • Consider captive breeding program")
    elif Ne_combined < 100:
        print("  ⚠️  WARNING: 50 < Ne < 100")
        print("  • Short-term survival okay, long-term risk high")
        print("  • Monitor closely, intervention likely needed")
        print("  • Increase breeding population if possible")
    elif Ne_combined < 500:
        print("  ℹ️  CONCERN: 100 < Ne < 500")
        print("  • Minimum for short-term viability achieved")
        print("  • Long-term evolutionary potential limited")
        print("  • Aim to increase Ne further")
    else:
        print("  ✓ ADEQUATE: Ne > 500")
        print("  • Sufficient for long-term evolutionary potential")
        print("  • Continue monitoring and maintain habitat")
    
    print("\n" + "="*70)
    print("THE 50/500 RULE (Franklin & Frankham)")
    print("="*70)
    print("\n  • Ne > 50: Avoid inbreeding depression (short-term)")
    print("  • Ne > 500: Maintain evolutionary potential (long-term)")
    print(f"\n  Your population: Ne = {Ne_combined:.0f}")
    if Ne_combined >= 500:
        print("  ✓ Exceeds both thresholds!")
    elif Ne_combined >= 50:
        print("  ⚠️  Exceeds 50 but below 500")
    else:
        print("  ⚠️  Below critical 50 threshold!")
    print("\n" + "="*70)

# Create interactive widget
interact(explore_ne_ratio,
         census_N=IntSlider(min=100, max=2000, step=100, value=1000,
                           description='Census N:', continuous_update=False),
         males=IntSlider(min=100, max=1000, step=50, value=400,
                        description='Males:', continuous_update=False),
         females=IntSlider(min=100, max=1000, step=50, value=600,
                          description='Females:', continuous_update=False),
         breeding_fraction=FloatSlider(min=0.1, max=1.0, step=0.1, value=0.5,
                                      description='Breeding %:', continuous_update=False),
         offspring_variance=FloatSlider(min=2, max=20, step=1, value=4,
                                       description='Offspring Vk:', continuous_update=False));
```

# Part 2: Estimation Methods

## How Do We Estimate Ne From Genetic Data?

We rarely know the true breeding structure of populations. But we can estimate Ne from genetic data!

---

## Method 1: Heterozygosity-Based

**Under drift:**
$$H_t = H_0\left(1 - \frac{1}{2N_e}\right)^t$$

**If you measure heterozygosity at two time points:**
$$N_e = \frac{t}{2\left(\frac{H_0 - H_t}{H_0}\right)}$$

---

## Method 2: Linkage Disequilibrium (LD)

**Key insight:** Small Ne creates more random LD!

$$E[r^2] = \frac{1}{3N_e} + \frac{1}{n}$$

Where:
- r² = LD between unlinked loci
- n = sample size

**Most commonly used** for single-sample Ne estimation.

---

## Method 3: Temporal (Fc Method)

**Compare allele frequencies at two time points:**

$$F_c = \frac{(p_1 - p_2)^2}{\bar{p}(1-\bar{p})}$$

$$N_e = \frac{t}{2F_c}$$

**Advantage:** Doesn't assume equilibrium

---

## Software Tools

- **NeEstimator:** Implements LD and temporal methods
- **GONE:** Uses LD with multiple time scales
- **VarEff:** Variance effective size
- **COLONY:** Sibship-based estimation

---

## Real Example: Labeo rohita Conservation

**Scenario:**
- Survey shows N ≈ 5000 adults in Mahanadi tributary
- Genetic data: 20 microsatellite loci
- Calculate Ne using LD method

**Hypothetical result:**
- Observed mean r² = 0.08
- Sample size n = 50
- After correction: Ne ≈ 200

**Interpretation:**
- Ne/N ≈ 0.04 (only 4% effective!)
- Why so low? Probably:
  - Few dominant males reproduce
  - High variance in reproductive success
  - Age structure (only some cohorts breed)

**Conservation action:**
- Ne = 200 is concerning (below 500 target)
- Need to protect habitat, reduce fishing pressure
- Consider hatchery supplementation carefully (might reduce Ne!)

# Summary & Key Takeaways

## What We Learned

### 1. The Concept
- Ne = size of ideal population with same drift rate
- Almost always Ne < N (often much less!)
- Ne/N ratio typically 0.1-0.5

### 2. Factors Reducing Ne
- Unequal sex ratios
- Variance in reproductive success
- Non-random mating
- Overlapping generations
- Fluctuating population size

### 3. Estimation Methods
- Heterozygosity decline (temporal)
- Linkage disequilibrium (single sample)
- Fc method (temporal)
- Multiple software tools available

### 4. Conservation Guidelines
- Ne > 50: Minimum for avoiding inbreeding
- Ne > 500: For long-term evolutionary potential
- Monitor Ne, not just census N!
- Genetic data reveals hidden problems
