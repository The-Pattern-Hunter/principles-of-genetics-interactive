# 🎲 Probability Foundations

## Understanding the Math Behind Mendel

Genetics is fundamentally about **probability**. Two key rules:

### 1. Product Rule (AND)

**Probability of independent events occurring together:**

$$P(A \text{ AND } B) = P(A) \times P(B)$$

**Example:** Probability of TT offspring
- P(T sperm) = 0.5
- P(T egg) = 0.5
- P(TT) = 0.5 × 0.5 = 0.25

### 2. Sum Rule (OR)

**Probability of mutually exclusive events:**

$$P(A \text{ OR } B) = P(A) + P(B)$$

**Example:** Probability of heterozygote (Tt)
- P(T sperm, t egg) = 0.25
- P(t sperm, T egg) = 0.25
- P(Tt) = 0.25 + 0.25 = 0.50

```python
# Probability Explorer

@interact(
    p_allele1=FloatSlider(min=0.1, max=0.9, step=0.05, value=0.5, 
                         description='P(Dominant):', style={'description_width': 'initial'}),
    show_tree=Checkbox(value=True, description='Show tree'),
    show_calc=Checkbox(value=True, description='Show calculations')
)
def probability_explorer(p_allele1, show_tree, show_calc):
    p_allele2 = 1 - p_allele1
    
    # Calculate probabilities
    p_AA = p_allele1 * p_allele1
    p_Aa = 2 * p_allele1 * p_allele2
    p_aa = p_allele2 * p_allele2
    
    fig, axes = plt.subplots(1, 2, figsize=(16, 7))
    
    # Tree diagram
    if show_tree:
        ax = axes[0]
        ax.set_xlim(0, 5)
        ax.set_ylim(0, 8)
        ax.axis('off')
        
        ax.plot(0.5, 4, 'ko', markersize=15)
        ax.text(0.2, 4, 'Start', fontsize=12, fontweight='bold')
        
        # Branches
        ax.plot([0.5, 1.5], [4, 5.5], 'b-', linewidth=3)
        ax.plot([0.5, 1.5], [4, 2.5], 'r-', linewidth=3)
        ax.text(1, 5, f'A ({p_allele1:.2f})', fontsize=11, fontweight='bold', color='blue')
        ax.text(1, 3, f'a ({p_allele2:.2f})', fontsize=11, fontweight='bold', color='red')
        
        # From A
        ax.plot([1.5, 3], [5.5, 6.5], 'b-', linewidth=2)
        ax.plot([1.5, 3], [5.5, 4.5], 'r-', linewidth=2)
        ax.plot(3, 6.5, 'go', markersize=12)
        ax.text(3.3, 6.5, f'AA: {p_AA:.3f}', fontsize=11, fontweight='bold', color='green')
        ax.plot(3, 4.5, 'yo', markersize=12)
        ax.text(3.3, 4.5, f'Aa: {p_allele1*p_allele2:.3f}', fontsize=11, fontweight='bold', color='orange')
        
        # From a
        ax.plot([1.5, 3], [2.5, 3.5], 'b-', linewidth=2)
        ax.plot([1.5, 3], [2.5, 1.5], 'r-', linewidth=2)
        ax.plot(3, 3.5, 'yo', markersize=12)
        ax.text(3.3, 3.5, f'aA: {p_allele2*p_allele1:.3f}', fontsize=11, fontweight='bold', color='orange')
        ax.plot(3, 1.5, 'ro', markersize=12)
        ax.text(3.3, 1.5, f'aa: {p_aa:.3f}', fontsize=11, fontweight='bold', color='red')
        
        ax.set_title('Probability Tree', fontsize=14, fontweight='bold')
    else:
        axes[0].text(0.5, 0.5, 'Enable tree', ha='center', va='center',
                    transform=axes[0].transAxes, fontsize=13)
        axes[0].axis('off')
    
    # Bar chart
    ax2 = axes[1]
    genotypes = ['AA', 'Aa', 'aa']
    probs = [p_AA, p_Aa, p_aa]
    colors = ['green', 'orange', 'red']
    
    bars = ax2.bar(genotypes, probs, color=colors, alpha=0.7, edgecolor='black', linewidth=2)
    ax2.set_ylabel('Probability', fontsize=13, fontweight='bold')
    ax2.set_title('Genotype Probabilities', fontsize=14, fontweight='bold')
    ax2.grid(True, alpha=0.3, axis='y')
    
    for bar, prob in zip(bars, probs):
        ax2.text(bar.get_x() + bar.get_width()/2., prob + 0.02,
                f'{prob:.3f}\n({prob*100:.1f}%)', ha='center', fontweight='bold')
    
    plt.tight_layout()
    plt.show()
    
    if show_calc:
        print("\n🎲 Calculations:")
        print("═" * 60)
        print(f"P(A) = {p_allele1:.3f}, P(a) = {p_allele2:.3f}")
        print(f"\nPRODUCT RULE:")
        print(f"  P(AA) = {p_allele1:.3f} × {p_allele1:.3f} = {p_AA:.4f}")
        print(f"  P(Aa) = 2 × {p_allele1:.3f} × {p_allele2:.3f} = {p_Aa:.4f}")
        print(f"  P(aa) = {p_allele2:.3f} × {p_allele2:.3f} = {p_aa:.4f}")
        print(f"\nSUM RULE:")
        print(f"  P(Dominant) = {p_AA:.4f} + {p_Aa:.4f} = {p_AA + p_Aa:.4f}")
        print("═" * 60)
```

# 📊 Dihybrid Cross - Two Genes!

## The Challenge

**Scenario:** Two genes segregating together
- Gene 1: Seed shape (R = round, r = wrinkled)
- Gene 2: Seed color (Y = yellow, y = green)

**Cross:** RrYy × RrYy

**Question:** What ratio emerges?

```python
# Dihybrid Cross Simulator

@interact(
    sample_size=IntSlider(min=16, max=1600, step=16, value=160, 
                         description='Offspring:', style={'description_width': 'initial'}),
    show_expected=Checkbox(value=True, description='Show 9:3:3:1')
)
def dihybrid_simulator(sample_size, show_expected):
    # Simulate
    phenotypes = []
    for _ in range(sample_size):
        r_alleles = np.random.choice(['R', 'r'], size=2)
        gene1 = 'R_' if 'R' in r_alleles else 'rr'
        
        y_alleles = np.random.choice(['Y', 'y'], size=2)
        gene2 = 'Y_' if 'Y' in y_alleles else 'yy'
        
        phenotypes.append(gene1 + gene2)
    
    # Count
    counts = Counter(phenotypes)
    order = ['R_Y_', 'R_yy', 'rrY_', 'rryy']
    observed = [counts.get(p, 0) for p in order]
    expected = [sample_size * r for r in [9/16, 3/16, 3/16, 1/16]]
    
    # Plot
    fig, ax = plt.subplots(figsize=(12, 7))
    
    labels = ['Round\nYellow', 'Round\nGreen', 'Wrinkled\nYellow', 'Wrinkled\nGreen']
    colors = ['lightgreen', 'lightyellow', 'lightcoral', 'lightgray']
    
    x = np.arange(len(labels))
    width = 0.35
    
    ax.bar(x - width/2, observed, width, label='Observed', 
           color=colors, alpha=0.8, edgecolor='black', linewidth=2)
    
    if show_expected:
        ax.bar(x + width/2, expected, width, label='Expected 9:3:3:1', 
               color='steelblue', alpha=0.5, edgecolor='black', linewidth=2)
    
    ax.set_ylabel('Count', fontsize=13, fontweight='bold')
    ax.set_title(f'Dihybrid Cross (n={sample_size})', fontsize=14, fontweight='bold')
    ax.set_xticks(x)
    ax.set_xticklabels(labels)
    ax.legend(fontsize=11)
    ax.grid(True, alpha=0.3, axis='y')
    
    for i, val in enumerate(observed):
        ax.text(i - width/2, val + max(observed)*0.02, f'{val}', 
               ha='center', fontweight='bold')
    
    plt.tight_layout()
    plt.show()
    
    # Summary
    if observed[3] > 0:
        ratio = [round(obs / observed[3], 1) for obs in observed]
    else:
        ratio = [0, 0, 0, 0]
    
    print("\n📊 Dihybrid Results:")
    print("═" * 70)
    print(f"Cross: RrYy × RrYy")
    print(f"Sample: {sample_size}")
    print(f"\nObserved ratio: {ratio[0]}:{ratio[1]}:{ratio[2]}:{ratio[3]}")
    print(f"Expected ratio: 9:3:3:1")
    print("\n💡 This is Independent Assortment!")
    print("   (3:1) × (3:1) = 9:3:3:1")
    print("═" * 70)
```

## Understanding 9:3:3:1

### Why This Ratio?

**Key:** Two genes = Two separate 3:1 crosses!

| Gene 1 | × | Gene 2 | = | Result | Fraction |
|--------|---|--------|---|--------|----------|
| 3 Round | × | 3 Yellow | = | 9 Round Yellow | 9/16 |
| 3 Round | × | 1 green | = | 3 Round green | 3/16 |
| 1 wrinkled | × | 3 Yellow | = | 3 wrinkled Yellow | 3/16 |
| 1 wrinkled | × | 1 green | = | 1 wrinkled green | 1/16 |

**Total: 9:3:3:1**

### Mendel's Law of Independent Assortment

**Law:** Genes for different traits segregate independently.

**Why?** Genes on different chromosomes (or far apart) assort independently during meiosis.
